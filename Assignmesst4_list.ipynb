### 12.Find the second greatest element in the list [89, 23, 24, 2, 55, 54, 64].

```python
l=[89, 23, 24, 2, 55, 54, 64]
for i in range(len(l)):
    m=l[i]
    for j in range(i+1,len(l)):
        if l[j]>l[i]:
            l[i],l[j]=l[j],l[i]
    if i==2:
        break
print("second greatest element is",l[1])
```

### 23.Find the frequency of each element in a list.

```python
l=[1, 2, 2, 3, 4, 1, 5, 2]
l1=[]
for i in range(len(l)):
    count=1
    if l[i] not in l1:
        for j in range(i+1,len(l)):
            if l[i]==l[j]:
                count+=1
        l1.append(l[i])
        print(l[i],"-->",count,"times")
```

### 36.Rearrange a list so that positive and negative numbers alternate

```python
l=[1, 2, -3, -4, 5, -6]
l1=[]
l2=[]
for i in l:
    if i>0:
        l1.append(i)
    else:
        l2.append(i)
res=[]
i=0
j=0
while(i<len(l1) and j<len(l2)):
    res.append(l1[i])
    res.append(l2[j])
    i+=1
    j+=1
print(res)
```

### 42.Find All Anagram Groups in a List of Words

```python
l=["eat", "tea", "tan", "ate", "nat", "bat"]
l1=[]
v=[]
for i in range(len(l)):
    a=[]
    if l[i] not in v:
        a.append(l[i])
        v.append(l[i])
        for j in range(i+1,len(l)):
            if sorted(l[i])==sorted(l[j]):
                a.append(l[j])
                v.append(l[j])
    if len(a)!=0:
        l1.append(a)
print(l1) 
```
