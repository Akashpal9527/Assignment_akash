# 11 calculate profit and loss percentage

```python
CP=int(input("enter CP:"))
SP=int(input("enter SP:"))
if SP>CP:
    profit=SP-CP
    per_profit=(profit/CP)*100
    print("per_profit=",per_profit)
elif CP>SP:
    loss=CP-SP
    per_loss=(loss/CP)*100
    print("per_loss=",per_loss)
else:
    print("not profit and not loss")
```
