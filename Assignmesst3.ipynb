# 37.Task: Count how many uppercase and lowercase letters are in a string.

```python
string=input("enter your string:")
count_u=0
count_l=0
for i in string:
    if i.isupper():
        count_u+=1
    elif i==" " or i.isdigit():
        continue
    else:
        count_l+=1
print("no of upper case=",count_u)
print("no of lower case=",count_l)
```

# 42.You are managing a simple banking system that tracks the balance at the end of each day over 10 days. Each day, the balance increases by 100 units starting from 100 on day 1, 200 on day 2, and so on. You want to print the current day’s balance along with the previous day’s balance. For day 1, the previous day’s balance is 0.


```python
Balance = 100
Previous_Day_Balance = 0
for i in range(1,11):
    if i==1:
        print(f"Day {i}:Balance=100,Previous Day Balance=0")
        Previous_Day_Balance+=100
    else:
        Balance+=100
        print(f"Day {i}:Balance={Balance},Previous Day Balance={Previous_Day_Balance}")
        Previous_Day_Balance=Balance
```
